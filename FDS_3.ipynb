# **ACTIVITY 2**

Write a programming code with a procedure to compute mean, mode, median, variance, standard deviation, and interquartile range of a given sample data without using **in-built functions**.


```python
import numpy as np

arr = x['temperature_c'].to_numpy()

np.set_printoptions(threshold=np.inf)
print(arr)
```

```python
total = 0

for element in arr:
    total += element

print("Sum:", total)
```

```python
mean=total/len(arr)
print("Mean:",mean)
```

```python
sarr = x['temperature_c'].to_numpy()

np.set_printoptions(threshold=np.inf)

def find_median(sarr):


    n = len(sarr)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if sarr[j] > sarr[j + 1]:

                sarr[j], sarr[j + 1] = sarr[j + 1], sarr[j]
                swapped = True

        if not swapped:
            break

    if n % 2 == 0:
        mid1 = n // 2 - 1
        mid2 = n // 2
        return (arr[mid1] + arr[mid2]) / 2
    else:
        return arr[n // 2]

print("Median:", find_median(arr))
```

```python
frequency_dict = {}
for item in arr:
    frequency_dict[item] = frequency_dict.get(item, 0) + 1

max_frequency = max(frequency_dict.values())

modes = [key for key, value in frequency_dict.items() if value == max_frequency]

print(f"Mode: {modes}")
```

```python
n = len(arr)

variance_sum = sum((x - mean) ** 2 for x in arr)
std = (variance_sum / n) ** 0.5

print("Variance:", variance_sum / n)
print("Standard Deviation:", std)
```

```python
import array

def gt_median(sarr, start_idx, end_idx):

    length = end_idx - start_idx
    mid = start_idx + (length // 2)

    if length % 2 == 1:
        return sarr[mid]
    else:
        return (sarr[mid - 1] + sarr[mid]) / 2.0


def iqr(sarr):

    mid = n // 2

    if n % 2 == 1:

        q1 = gt_median(sarr, 0, mid)
        q3 = gt_median(sarr, mid + 1, n)
    else:

        q1 = gt_median(sarr, 0, mid)
        q3 = gt_median(sarr, mid, n)

    return q3 - q1


iqr_result = iqr(sarr)

print("IQR:", iqr_result)
```

# **ACTIVITY - 3**

Compute and analyze the grouped data to find out frequency, mean, mode, median, standard deviation, variance, and interquartile range of data.


```python
carr = x['relative_humidity'][11640:11760].to_numpy()
print(carr)
```

```python
n = len(carr)

k = 1 + 3.322 * np.log10(n)

print(f"Number of data points (n): {n}")
print(f"Number of bins (k) using Sturges' formula: {k:.0f}")
```

```python
min_humd = np.min(carr)
max_humd = np.max(carr)
data_range = max_humd - min_humd

num_bins = int(round(k))

bin_width = data_range / num_bins

print(f"Minimum temperature: {min_humd}")
print(f"Maximum temperature: {max_humd}")
print(f"Data range: {data_range}")
print(f"Calculated number of bins (k): {num_bins}")
print(f"Bin width: {bin_width}")
```

```python
frequencies, bin_edges = np.histogram(carr, bins=num_bins)

bin_labels = []
for i in range(len(bin_edges) - 1):
    bin_labels.append(f"{bin_edges[i]:.2f} - {bin_edges[i+1]:.2f}")

frequency_df = pd.DataFrame({
    'Bin Range': bin_labels,
    'Frequency': frequencies
})

display(frequency_df)
```

```python
bin_midpoints = []
for i in range(len(bin_edges) - 1):
    midpoint = (bin_edges[i] + bin_edges[i+1]) / 2
    bin_midpoints.append(midpoint)

frequency_df['Bin Midpoint'] = bin_midpoints

display(frequency_df)
```

```python
sum_midpoint_frequency = (frequency_df['Bin Midpoint'] * frequency_df['Frequency']).sum()

total_data_points = n

mean_from_table = sum_midpoint_frequency / total_data_points

print(f"Sum of sigma: {sum_midpoint_frequency:}")

print(f"Mean from Frequency Table: {mean_from_table:}")
```

```python
modal_class_idx = frequency_df['Frequency'].idxmax()

mode_from_table = frequency_df.loc[modal_class_idx, 'Bin Midpoint']

print(f"Modal Class: {frequency_df.loc[modal_class_idx, 'Bin Range']}")
print(f"Mode from Frequency Table: {mode_from_table:}")
```

```python
frequency_df['Cumulative Frequency'] = frequency_df['Frequency'].cumsum()

median_position = n / 2
median_class_idx = frequency_df[frequency_df['Cumulative Frequency'] >= median_position].index[0]
median_class_info = frequency_df.loc[median_class_idx]

L = bin_edges[median_class_idx]
cf = frequency_df.loc[median_class_idx - 1, 'Cumulative Frequency'] if median_class_idx > 0 else 0
f = median_class_info['Frequency']
w = bin_width

median_from_table = L + (((n/2) - cf) / f) * w

print(f"Median Position (n/2): {median_position}")
print(f"Median Class: {median_class_info['Bin Range']}")
print(f"Median from Frequency Table: {median_from_table:}")

display(frequency_df)
```

```python
frequency_df['(Midpoint - Mean)^2 * Freq'] = (

     (frequency_df['Bin Midpoint'] - mean_from_table)**2) * frequency_df['Frequency']


sum_sq_diff_freq = frequency_df['(Midpoint - Mean)^2 * Freq'].sum()

variance_from_table = sum_sq_diff_freq / n

std_from_table = np.sqrt(variance_from_table)

print(f"Estimated Variance from Frequency Table: {variance_from_table:}")
print(f"Estimated Standard Deviation from Frequency Table: {std_from_table:}")

display(frequency_df.drop(columns=['(Midpoint - Mean)^2 * Freq']))
```

```python
q1_position = n / 4

q1_class_idx = frequency_df[frequency_df['Cumulative Frequency'] >= q1_position].index[0]
q1_class_info = frequency_df.loc[q1_class_idx]

L_q1 = bin_edges[q1_class_idx]
cf_q1 = frequency_df.loc[q1_class_idx - 1, 'Cumulative Frequency'] if q1_class_idx > 0 else 0
f_q1 = q1_class_info['Frequency']
w_q1 = bin_width

q1_from_table = L_q1 + (((n/4) - cf_q1) / f_q1) * w_q1


q3_position = 3 * n / 4
q3_class_idx = frequency_df[frequency_df['Cumulative Frequency'] >= q3_position].index[0]
q3_class_info = frequency_df.loc[q3_class_idx]

L_q3 = bin_edges[q3_class_idx]
cf_q3 = frequency_df.loc[q3_class_idx - 1, 'Cumulative Frequency'] if q3_class_idx > 0 else 0
f_q3 = q3_class_info['Frequency']
w_q3 = bin_width

q3_from_table = L_q3 + (((3*n/4) - cf_q3) / f_q3) * w_q3

iqr_from_table = q3_from_table - q1_from_table

print(f"Q1 Position (n/4): {q1_position}")
print(f"Q1 Class: {q1_class_info['Bin Range']}")
print(f"Q1 from Frequency Table: {q1_from_table:}")

print(f"Q3 Position (3n/4): {q3_position}")
print(f"Q3 Class: {q3_class_info['Bin Range']}")
print(f"Q3 from Frequency Table: {q3_from_table:}")

print(f"IQR from Frequency Table: {iqr_from_table:}")
```
